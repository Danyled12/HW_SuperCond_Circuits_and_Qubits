- Выбираем две функции:

- Константная: $f(x_1,x_2)=0$;

- оракул для неё $U_f = I $(посчитано ручками на бумаге);


- Сбалансированная: $f(x_1,x_2)=x_0$;

- оракул для неё $ U_f = CX (x_1,y) $(посчитано ручками на бумаге)



```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit_ibm_runtime.fake_provider import FakeManilaV2
import matplotlib.pyplot as plt

SHOTS = 4096
```

```python
# Пусть kind - принимает либо constant, либо balanced (иначе - ошибка)
# Тогда функция будет строить соответствующий оракул U_f
def make_oracle(kind: str) -> QuantumCircuit:
    qc = QuantumCircuit(3, name=f"Uf_{kind}")
    if kind == "constant": # так как U_f(const) = I, то ничего не должно меняться, пропускаем как есть
        pass
    elif kind == "balanced": # так как U_f(balanced) = CX(x1,y), то, соответственно, применяем CX
        qc.cx(0, 2)
    else:
        raise ValueError("Error") #защита от других аргументов
    return qc

#Построим саму схему алгоритма
def make_dj_circuit(kind: str) -> QuantumCircuit:
    qc = QuantumCircuit(3, 2) # в результате будем измерять 2 бита, поэтому объявляем столько классических


    qc.x(2) # кубит y
    qc.h(0) # кубит x_1
    qc.h(1) # кубит x_2
    qc.h(2)
    # первый слой - применяем ко всем оператор Адамара, а к q_3 - X, так как изначально он в состоянии \0>
    
    # Uf
    qc.compose(make_oracle(kind), inplace=True)

    qc.h(0)
    qc.h(1)

    qc.measure(0, 0)
    qc.measure(1, 1)
    return qc

constant_circuit = make_dj_circuit("constant")
balanced_circuit = make_dj_circuit("balanced")

constant_circuit.draw("text")
```

```python
print("Constant oracle (nothing change, so its empty):")
print(make_oracle("constant").draw("text"))
print("Balanced oracle:")
print(make_oracle("balanced").draw("text"))
print("Deutsch-Jozsa, constant:")
print(constant_circuit.draw("text"))
print("Deutsch-Jozsa, balanced:")
print(balanced_circuit.draw("text"))
```

```python
# Сначала посмотрим, что получается в случае без шума (идеальном)
ideal_backend = AerSimulator()

# Затем смотрим результат на "реальной" машине с имитацией физического шума
fake_backend = FakeManilaV2()
noisy_backend = AerSimulator.from_backend(fake_backend)


def run_counts(circuit, backend, shots=SHOTS):
    compiled = transpile(circuit, backend)
    result = backend.run(compiled, shots=shots).result()
    return result.get_counts()

results = {}
for name, circuit in [("constant", constant_circuit), ("balanced", balanced_circuit)]:
    results[(name, "ideal")] = run_counts(circuit, ideal_backend)
    results[(name, "noisy")]= run_counts(circuit, noisy_backend)

results
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharey=False)
for row, name in enumerate(["constant", "balanced"]):
    for col, mode in enumerate(["ideal", "noisy"]):
        counts = results[(name, mode)]
        ax = axes[row, col]
        keys = sorted(counts)
        vals = [counts[k] for k in keys]
        ax.bar(keys, vals)
        ax.set_title(f"{name} — {mode}")
        ax.set_xlabel("Измеренный битовый результат c1c0")
        ax.set_ylabel("Число отсчётов")
        ax.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
def summarize(name, mode, counts):
    # For our chosen functions the ideal result is known analytically:
    # constant f=0 -> 00; balanced f=x0 -> 01 (q0 is the rightmost bit).
    expected = "00" if name == "constant" else "01"
    total = sum(counts.values())
    correct = counts.get(expected, 0)
    error = total - correct
    return {
        "function": name,
        "mode": mode,
        "expected": expected,
        "correct_count": correct,
        "error_count": error,
        "correct_fraction": correct / total,
        "error_fraction": error / total,
        "top_results": sorted(counts.items(), key=lambda kv: kv[1], reverse=True)[:5],
    }

summary = [summarize(name, mode, results[(name, mode)])
           for name in ["constant", "balanced"]
           for mode in ["ideal", "noisy"]]
for row in summary:
    print(row)
```

Обсуждение 

Идеальный случай:
В идеальном запуске алгоритм Дойча—Йожи должен дать:

- для **константной** функции только `00` на двух входных кубитах;
- для **сбалансированной** функции только `01` для выбранного $f(x_0,x_1)=x_0$.

При чтении строки `01` важно помнить соглашение Qiskit: крайний правый бит — кубит 0. Следовательно, `01` означает $q_0=1$, $q_1=0$.

В шумном запуске появляются дополнительные битовые строки. Доля правильного исхода становится меньше 1, потому что модель `FakeManilaV2` учитывает ошибки реального устройства (в том числе ошибки однокубитных/двухкубитных операций и измерений, в зависимости от версии модели).

Для классификации достаточно посмотреть на **доминирующий** результат:

- пик около `00` указывает на константный случай;
- пик около `01` указывает на выбранный сбалансированный случай.

При достаточно сильном шуме распределения могут перекрываться, поэтому одна отдельная выборка уже не даёт абсолютной гарантии. В этом ноутбуке дополнительно печатаются `correct_fraction` и `error_fraction`, чтобы отдельно увидеть, какая часть отсчётов ушла от идеального результата.
